```python
df = df.rename(columns ={'host id':'host_id','host name':'host_name','neighbourhood group':'neighbourhood_group','room type':'room_type',
                        'Construction year':'Construction_year','service fee':'service_fee','minimum nights':'minimum_nights',
                         'number of reviews':'number_of_reviews','last review':'last_review','reviews per month':'reviews_per_month',
                         'review rate number':'review_rate_number','calculated host listings count':'calculated_host_listings_count',
                         'availability 365':'availability_365'})
                                                                                                                                                
                                                                                                                                                
                                                                                                                                                                                                                                                                          
```

```python
df.columns
```

```python
df
```

```python
df.neighbourhood_group.unique()
```

```python
df.neighbourhood_group = df.neighbourhood_group.replace('brookln','Brooklyn')
df.neighbourhood_group = df.neighbourhood_group.replace('manhatan','Manhattan')
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
df.NAME = df.NAME.fillna('unknown')
df.host_identity_verified = df.host_identity_verified.fillna('unknown')
df.host_name = df.host_name.fillna('unknown')
df.neighbourhood_group = df.neighbourhood_group.fillna('unknown')
df.neighbourhood = df.neighbourhood.fillna('unknown')
df.country = df.country.fillna('unknown')
df.instant_bookable = df.instant_bookable.fillna('unknown')
df.cancellation_policy = df.cancellation_policy.fillna('unknown')
df.Construction_year = df.Construction_year.fillna(2012)
df.price = df.price.fillna(625)
df.service_fee = df.service_fee.fillna(125)
df.minimum_nights = df.minimum_nights.fillna(8.135)
df.number_of_reviews = df.number_of_reviews.fillna(27)
df.last_review = df.last_review.fillna(0)
df.reviews_per_month = df.reviews_per_month.fillna(1.37)
df.review_rate_number =  df.review_rate_number.fillna(3.27)
df.calculated_host_listings_count = df.calculated_host_listings_count.fillna(7.93)
df.availability_365 = df.availability_365.fillna(141)
df.license = df.license.fillna('unknown')
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2,3,1)
n_l= df.groupby('neighbourhood')['calculated_host_listings_count'].mean().nlargest(10)
sb.barplot(x=n_l.index, y=n_l.values, hue=n_l.values, palette='Spectral')
plt.xticks(rotation = 90)
plt.title('Neighbourhood by Listening', color='red')
plt.ylabel('calculated_host_listings_count')


plt.subplot(2,3,2)
n_p= df.groupby('neighbourhood')['price'].mean().nlargest(10)
sb.barplot(x=n_p.index, y=n_p.values, hue=n_p.values, palette='husl')
plt.xticks(rotation = 90)
plt.title('Neighbourhood by Price', color='red')
plt.ylabel('price')



plt.subplot(2,3,3)
p = df.price.value_counts().nlargest(10)
sb.barplot(x=p.index, y=p.values, hue=p.values, palette='muted')
plt.title('Price Distribution', color='red')
plt.ylabel('count')


plt.subplot(2,3,4)
r_t = df.room_type.value_counts().nlargest(10)
sb.barplot(x=r_t.index, y=r_t.values, hue=r_t.values, palette='colorblind')
plt.title('Room Type Distribution', color='red')
plt.xticks(rotation = 90)
plt.ylabel('count')


plt.subplot(2,3,5)
a_y = df.groupby('Year')['availability_365'].mean().nlargest(10)
sb.barplot(x=a_y.index,y=a_y.values,hue=a_y.values,palette='mako')
plt.title('Availability Over the Year', color='red')
plt.xticks(rotation = 90)
plt.ylabel('availability_365')

plt.tight_layout(pad=4)
plt.show()
```
